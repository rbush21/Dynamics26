$f_0=0$

```python
import numpy as np
from scipy import integrate
import matplotlib.pyplot as plt
#Idea inspired from
#https://www.geeksforgeeks.org/maths/program-for-picards-iterative-method-computational-mathematics/

# 1. Define grid (since bounds are the same, x and y share this grid)
x = np.linspace(-1, 1, 1000)
y = np.linspace(-1, 1, 1000)

# 2. Define the operator function
def T(f, lambda_var):
    # Inside integral term
    # x needs to be passed as a 2D array in order to function and be considered
    # as a constant in the integral and compare all values of x, y
      # Source: https://numpy.org/doc/stable/user/basics.broadcasting.html
    integral_term = 1.0 / (1.0 + (x[:, None] - y)**2)

    # Integrate over y (axis=1) using trapezoid numerical integration technique
    integral_val = integrate.trapezoid(f * integral_term, y, axis=1)

    # Return the new function value as a function of x
    return np.sin(2 * np.pi * x) + lambda_var * integral_val

#(I know there is a more efficient way to do this, but I don't feel like
#fussing around with Python any more than I have to)

#Define universal values used for all loops
iterations = 100

#-------------Lambda=0.4--------------------------

#Define lambda (which changes)
lambda_val_1 = 0.4
f_1 = 0 # starting function f_0

for i in range(iterations):
    # Compute next iteration
    f_next_it_1 = T(f_1, lambda_val_1)

    f_1 = f_next_it_1

# Plot final f vs x
plt.plot(x, f_1, label=f'$\lambda = {lambda_val_1}$', color='blue', linewidth=2)


#-------------Lambda=2/pi--------------------------

##Define lambda (which changes)
lambda_val_2 = 2/np.pi
f_2 = 0 # starting function f_0

for i in range(iterations):
    # Compute next iteration
    f_next_it_2 = T(f_2, lambda_val_2)

    f_2 = f_next_it_2

# Plot final f vs x
plt.plot(x, f_2, label=r'$\lambda = \frac{2}{\pi}$', color='red', linewidth=2)

#-------------Lambda=1--------------------------

##Define lambda (which changes)
lambda_val_3 = 1
f_3 = 0 # starting function f_0

for i in range(iterations):
    # Compute next iteration
    f_next_it_3 = T(f_3, lambda_val_3)

    f_3 = f_next_it_3

# Plot final f vs x
plt.plot(x, f_3, label=f'$\lambda = {lambda_val_3}$', color='green', linewidth=2)

#-------------Lambda=1.05--------------------------

##Define lambda (which changes)
lambda_val_4 = 1.05
f_4 = 0 # starting function f_0

for i in range(iterations):
    # Compute next iteration
    f_next_it_4 = T(f_4, lambda_val_4)

    f_4 = f_next_it_4

# Plot final f vs x
plt.plot(x, f_4, label=f'$\lambda = {lambda_val_4}$', color='black', linewidth=2)


# Add labels + title
plt.title('Numerical Investigation of Fixed Point, where $f_0(x)=0$', fontsize=14)
plt.xlabel('x', fontsize=12)
plt.ylabel('f(x)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)

# Display the plot
plt.show()
```
