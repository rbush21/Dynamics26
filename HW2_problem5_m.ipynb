```python
import matplotlib.pyplot as plt
from scipy.integrate import odeint
import numpy as np
from mpmath import nsum, inf, cos
#idea for this is from :
#https://mpmath.org/doc/current/calculus/sums_limits.html$0

#Define x values
x = np.linspace(-np.pi, np.pi, 1000)

#Define W(x)

def W(x, n, alpha):

  result = nsum(lambda n: 2**(-n*alpha) * cos((2**n)*x), [0, n])
  return result

#define vector to test different values of alpha
alpha = [0.25, 0.5, 0.75]

#define W, where n=1,000 (a large number of terms)

#have to define arrays to store the

alpha_0_wfunc = np.zeros_like(x)

alpha_1_wfunc = np.zeros_like(x)

alpha_2_wfunc = np.zeros_like(x)


for x_i in range(len(x)):
  for i in range(len(alpha)):
    if i == 0:
      alpha_0_wfunc[x_i] = W(x[x_i], 1000, alpha[i])
    if i == 1:
      alpha_1_wfunc[x_i] = W(x[x_i], 1000, alpha[i])
    if i == 2:
      alpha_2_wfunc[x_i] = W(x[x_i], 1000, alpha[i])


#plot our results
plt.plot(x, alpha_0_wfunc, label = r'$\alpha=0.25$')

plt.plot(x, alpha_1_wfunc, label = r'$\alpha=0.5$')

plt.plot(x, alpha_2_wfunc, label = r'$\alpha=0.75$')


# insert the title
plt.title(f'continuous but not lipschitz on $[-\pi,\pi]$, $n=1,000$', fontsize=14)
# set the axis labels
plt.xlabel('x',fontsize=12)
plt.ylabel('W(x)',fontsize=12)
# adjust the font size of x and y axes
#plt.tick_params(axis='both', which='major', labelsize=14)
# insert legend
plt.legend(fontsize=14, loc='center left', bbox_to_anchor=(1, 0.5))
plt.show()
```
